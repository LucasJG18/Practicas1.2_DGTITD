```python
import pandas as pd

# Definimos la ruta del archivo Excel y lo cargamos en un DataFrame usando el motor 'openpyxl'.
print("Leyendo Registro Administrativo 2009-2024...")
ruta_excel = 'data/registro-administrativo-historico_2009-2024-inicio.xlsx'
df = pd.read_excel(ruta_excel, sheet_name=0, engine='openpyxl')

# 1.  Obtenemos el número total de filas y columnas
print("\n1. Dimensiones:", df.shape)

# 2. Extraemos y mostramos los nombres de las primeras 10 columnas para ver cómo vienen reportadas.
print("\n--- Nombres de las primeras columnas ---")
print(df.columns.tolist()[:10])

# 3. Calculamos la cantidad de valores vacíos por cada columna y mostramos el Top 15 en orden descendente.
print("\n2. Top 15 columnas con nulos:")
print(df.isnull().sum().sort_values(ascending=False).head(15))

# 4. ]Evaluamos la cantidad de registros para las variables categóricas clave (Sostenimiento y Nivel de Educación).
# Usamos un bloque try-except para evitar errores en caso de que los nombres varíen en la fuente.
try:
    print("\n3. Distribución por Sostenimiento:")
    print(df['Sostenimiento'].value_counts())
    
    print("\n4. Distribución por Nivel de Educación:")
    print(df['Nivel Educacion'].value_counts())
except KeyError:
    print("\nOjo: Al parecer las columnas tienen nombres ligeramente distintos en el Excel")
```
